# Human Vitals Anomaly Detection - Public Workflow

This notebook documents the public-safe workflow used for the thesis project. It is designed for GitHub review and avoids uploading raw patient-level working files.

## Project Goal

Detect unusual vital-sign patterns using an unsupervised Isolation Forest workflow, then explain the modelling approach in a way that can be reviewed by analysts and recruiters.

```python
import pandas as pd
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler
```

## Load Public-Safe Sample

The full thesis used 200,020 records. The GitHub version uses a small public-safe sample and aggregate outputs so the project can be reviewed without exposing sensitive working data.

```python
sample_path = '../data/public_sample_human_vitals.csv'
df = pd.read_csv(sample_path)
df.head()
```

## Prepare Features

Numeric vital-sign fields are selected and scaled before anomaly detection. Scaling helps the model compare fields measured on different ranges.

```python
feature_columns = [column for column in df.columns if df[column].dtype != 'object']
features = df[feature_columns].dropna()
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

## Train Isolation Forest

Isolation Forest is used because the project focuses on unsupervised anomaly detection, where labelled abnormal cases may not be available.

```python
model = IsolationForest(n_estimators=200, contamination=0.05, random_state=42)
model.fit(scaled_features)
scores = features.copy()
scores['anomaly_label'] = model.predict(scaled_features)
scores['anomaly_score'] = model.decision_function(scaled_features)
scores['is_anomaly'] = scores['anomaly_label'].map({1: 0, -1: 1})
scores.head()
```
